```python
#import the libraries needed

import pandas as pd
```

```python
#read AE001.csv dataset and create dataframe

ae_df = pd.read_csv("AE001.csv")
```

```python
#read SV001.csv dataset and create dataframe
sv_df = pd.read_csv("SV001.csv")
```

```python
#Create a list of column names for each dataframe

ae_cols = ae_df.columns.tolist()
sv_cols = sv_df.columns.tolist()
print("AE Columns:", ae_cols)
print("SV Columns:", sv_cols)
```

```python
#the number of unique patients that experienced adverse event

unique_ae_patients = ae_df["Subject"].nunique()
print("Unique AE Patients:", unique_ae_patients)
```

```python
#the number of unique patients that had a study visit
unique_sv_patients = sv_df["Subject"].nunique()
print("Unique SV Patients:", unique_sv_patients)
```

```python
#Did all subjects in the study experience an adverse event?
#Please explain

all_experienced_ae = unique_ae_patients == unique_sv_patients
print("did all subjects experience an AE?", all_experienced_ae)


# No, only 31 unique patients reported an adverse event out of the 727 total unique patients who had study visits.
```

```python
#number of unique sites

unique_sites = sv_df["Site"].nunique()
print("Unique Sites:", unique_sites)
```

```python
#Identify the scheduled visits in this study

scheduled_visits = sv_df["Visits"].unique()
print("Visit Types:", scheduled_visits)
```

```python
#identify the # of AEs per patient in the whole study

aes_per_patient = ae_df["Subject"].value_counts()
print(aes_per_patient)
```

```python
#identify the # of AEs per site in the whole study
#How would you present this information to the study team?

aes_per_site = ae_df["Site"].value_counts()
print(aes_per_site)


# The raw data could be presented in a simple bar chart. However, presenting raw counts is misleading because sites have different enrollment numbers. The best way to present this is as a normalized percentage in a table or sorted bar chart or pie chart.
```

```python
#Any other considerations for this data? 
#Should normalizing the data be considered in this study?
#Why or why not?
#if yes, how would you go about this?

'''
Yes, normalizing the data is necessary. Larger sites will have more AEs simply because they have more patients. To fairly compare safety performance across sites, we must normalize the data by calculating the AE rate per capita for each site.

How to do it:
(1) Count the number of unique subjects per site from the sv_df.
()2 Count the number of AEs per site from the ae_df.
(3) Divide the AE count by the subject count for each site and multiply by 100 to get the percentage.
'''


subjects_per_site = sv_df.groupby("Site")["Subject"].nunique()
normalized_ae_rate = (aes_per_site / subjects_per_site).dropna() * 100
print(normalized_ae_rate.sort_values(ascending=False))
```
